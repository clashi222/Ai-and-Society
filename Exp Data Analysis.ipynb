```python
import os

base_dir = "/kaggle/input/datasets/abdelghaniaaba/wildfire-prediction-dataset"
print(os.listdir(base_dir)) 
```

```python
import os

splits = ["train", "valid", "test"]
for s in splits:
  s_path = os.path.join(base_dir, s)
  if os.path.exists(s_path):
    for c in os.listdir(s_path):
      c_path = os.path.join(s_path, c)
      if os.path.isdir(c_path):
        print(s, c, len(os.listdir(c_path))) 
```

```python
import os
import matplotlib.pyplot as plt

train_path = os.path.join(base_dir, "train")
classes = os.listdir(train_path)
counts = [len(os.listdir(os.path.join(train_path, c))) for c in classes]

plt.figure(figsize=(6, 4))
plt.bar(classes, counts, color=["coral", "skyblue"])
plt.title("Train Set Class Distribution")
plt.ylabel("Image Count")
plt.show() 
```

```python
import os
import matplotlib.pyplot as plt
from PIL import Image

fig, axes = plt.subplots(1, 2, figsize=(8, 4))
for i, c in enumerate(classes):
  c_dir = os.path.join(train_path, c)
  img_name = os.listdir(c_dir)[0]
  img = Image.open(os.path.join(c_dir, img_name))
  axes[i].imshow(img)
  axes[i].set_title(c)
  axes[i].axis("off")
plt.show() 
```

```python
import os
from PIL import Image

widths, heights = [], []
for c in classes:
  c_dir = os.path.join(train_path, c)
  for img_name in os.listdir(c_dir)[:50]:
    img_path = os.path.join(c_dir, img_name)
    with Image.open(img_path) as img:
      w, h = img.size
      widths.append(w)
      heights.append(h)

print("Average Width:", sum(widths) / len(widths))
print("Average Height:", sum(heights) / len(heights)) 
```

```python
import os

total_images = 0
for s in splits:
  s_path = os.path.join(base_dir, s)
  if os.path.exists(s_path):
    for root, dirs, files in os.walk(s_path):
      total_images += len(files)

print("Total dataset images found:", total_images)
```
